# Class 4. Object Model & Type System

이름과 객체의 관계를 `type()`, `id()`, 참조 횟수로 확인한다. 가변 객체와 불변 객체를 비교하고, 복사와 함수 기본 인자에서 생기는 차이를 실험한다. 이어서 Python의 타입 체계와 Type Hint를 다룬다.


## 1. Object Model

### 1.1 Everything is Object

Python에서는 숫자와 문자열뿐 아니라 함수, 클래스, 타입도 객체다. 객체는 값과 타입을 가지며, 타입에 따라 사용할 수 있는 연산과 메서드가 정해진다.

변수는 객체에 붙인 이름이다. 아래 예제에서 숫자, 문자열, 함수, 클래스에 `type()`을 적용해 보자.


````python
# Everything is Object 증명
print("=== Everything is Object ===")
print()

# 숫자도 객체
x = 42
print(f"숫자 42:")
print(f"  type: {type(x)}")
print(f"  메서드: {[m for m in dir(x) if not m.startswith('_')][:5]}")
print()

# 문자열도 객체
s = "hello"
print(f"문자열 'hello':")
print(f"  type: {type(s)}")
print(f"  메서드: {[m for m in dir(s) if not m.startswith('_')][:5]}")
print()

# 함수도 객체


def greet():
  return "Hello"


print(f"함수 greet:")
print(f"  type: {type(greet)}")
print(f"  함수를 변수에 할당: ", end="")
f = greet
print(f"{f()}")
print()

# 타입도 객체
print(f"타입 int:")
print(f"  type(int): {type(int)}")
print(f"  isinstance(int, object): {isinstance(int, object)}")
````

### 1.2 Object의 속성과 참조 횟수

객체를 살펴볼 때는 Type, Value, Identity를 구분한다. 여기에 CPython의 메모리 관리와 관련된 Reference Count도 함께 확인한다.

| 항목 | 확인할 내용 |
| --- | --- |
| Type | 어떤 연산과 메서드를 지원하는 객체인가? |
| Value | 객체가 담고 있는 데이터는 무엇인가? |
| Identity | 같은 객체인가, 별개의 객체인가? |
| Reference Count | 객체를 가리키는 참조가 얼마나 있는가? |


#### 1) Type (타입)

Type은 객체의 종류이며, 사용할 수 있는 연산과 메서드를 결정한다. `type()`으로 확인한다. `int`, `str`, `list` 같은 타입도 객체이므로 이들에도 `type()`을 적용할 수 있다.


````python
# Type 확인
print("=== Object의 Type ===")
print()

x = 42
s = "Hello"
lst = [1, 2, 3]

print(f"x = 42:        type = {type(x)}")
print(f"s = 'Hello':   type = {type(s)}")
print(f"lst = [1,2,3]: type = {type(lst)}")
print()

# Type은 변경 불가
print("Type은 생성 후 변경 불가:")
x = 42
print(f"  x = 42,      type: {type(x)}")
x = "Hello"  # 새로운 객체 생성 (타입 변경 아님!)
print(f"  x = 'Hello', type: {type(x)}")
print("  → 타입이 바뀐 게 아니라 다른 객체를 가리킴!")
````

#### 2) Value (값)

Value는 객체가 담고 있는 데이터다. Mutable 객체는 같은 객체를 유지하면서 내용을 바꿀 수 있다. 반면 Immutable 객체의 값은 바뀌지 않는다. `list`, `dict`, `set`과 `int`, `str`, `tuple`을 비교해 보자.


````python
# Value 확인
print("=== Object의 Value ===")
print()

# Mutable: 값 변경 가능
lst = [1, 2, 3]
print(f"Mutable (list):")
print(f"  원본: {lst}, id: {id(lst)}")
lst[0] = 10  # 값 변경
print(f"  변경: {lst}, id: {id(lst)}")
print(f"  → 같은 객체의 값만 변경됨 (id 동일)")
print()

# Immutable: 값 변경 불가
x = 42
print(f"Immutable (int):")
print(f"  원본: {x}, id: {id(x)}")
x = 43  # 새 객체 생성
print(f"  '변경': {x}, id: {id(x)}")
print(f"  → 새로운 객체 생성 (id 변경)")
````

#### 3) Identity (정체성)

Identity는 객체가 살아 있는 동안 유지되는 고유 식별 정보이며, `id()`로 확인한다. CPython에서 `id()`는 객체의 메모리 주소에 해당한다.

- `is`: 같은 객체를 가리키는지 비교한다.
- `==`: 두 객체의 값이 같은지 비교한다.


````python
# Identity 확인
print("=== Object의 Identity ===")
print()

# 같은 값, 다른 객체
a = [1, 2, 3]
b = [1, 2, 3]
print(f"a = [1, 2, 3], id: {id(a)}")
print(f"b = [1, 2, 3], id: {id(b)}")
print(f"a == b: {a == b}  (값 비교)")
print(f"a is b: {a is b}  (정체성 비교)")
print()

# 같은 객체 가리키기
c = a
print(f"c = a")
print(f"c의 id: {id(c)}")
print(f"a is c: {a is c}  (같은 객체!)")
print()

# 특수 케이스: 작은 정수는 캐싱
x = 10
y = 10
print(f"x = 10, y = 10")
print(f"x is y: {x is y}  (작은 정수는 캐싱됨)")

x = 1000
y = 1000
print(f"x = 1000, y = 1000")
print(f"x is y: {x is y}  (큰 정수는 다른 객체)")
````

#### 4) Reference Count (참조 횟수)

CPython은 객체의 참조 횟수를 메모리 관리에 사용한다. 변수뿐 아니라 컨테이너의 요소 등도 객체를 참조할 수 있다. `sys.getrefcount()`로 관찰하되, 함수 호출에 전달하는 참조도 측정값에 포함된다는 점에 주의한다.

아래 예제에서는 같은 객체에 이름을 하나 더 붙이거나 이름을 삭제할 때 참조 횟수가 어떻게 달라지는지 확인한다.


````python
# Reference Count 확인
import sys

print("=== Reference Count ===")
print()

x = [1, 2, 3]
print(f"x = [1, 2, 3]")
print(f"refcount: {sys.getrefcount(x) - 1}")  # -1: getrefcount 자체의 참조 제외
print()

y = x  # 참조 증가
print(f"y = x")
print(f"refcount: {sys.getrefcount(x) - 1}")
print()

z = x  # 참조 증가
print(f"z = x")
print(f"refcount: {sys.getrefcount(x) - 1}")
print()

del y  # 참조 감소
print(f"del y")
print(f"refcount: {sys.getrefcount(x) - 1}")
print()

del z  # 참조 감소
print(f"del z")
print(f"refcount: {sys.getrefcount(x) - 1}")
````

````python
print(sys.getrefcount(int))
````

### 1.3 First-class Object

함수, 클래스, 모듈도 다른 값처럼 변수에 할당하고, 인자로 전달하고, 반환값이나 자료구조의 요소로 사용할 수 있다. 이를 일급 객체(First-class Object)라고 한다. 함수를 다른 함수에 전달하는 성질은 고차 함수, Decorator, Closure의 바탕이 된다.


````python
# First-class Function
print("=== First-class Function ===")
print()


def square(x):
  return x ** 2


def cube(x):
  return x ** 3


# 1. 변수에 할당
f = square
print(f"1. 변수에 할당: f = square")
print(f"   f(5) = {f(5)}")
print()

# 2. 함수의 인자로 전달


def apply(func, value):
  return func(value)


print(f"2. 인자로 전달: apply(square, 5)")
print(f"   결과: {apply(square, 5)}")
print()

# 3. 함수의 반환값


def get_operation(op):
  if op == 'square':
    return square
  else:
    return cube


operation = get_operation('square')
print(f"3. 반환값: operation = get_operation('square')")
print(f"   operation(5) = {operation(5)}")
print()

# 4. 자료구조에 저장
operations = [square, cube]
print(f"4. 리스트에 저장: operations = [square, cube]")
print(f"   operations[0](5) = {operations[0](5)}")
print(f"   operations[1](5) = {operations[1](5)}")
````

## 2. Mutability

### 2.1 Mutable vs Immutable

Mutability는 객체를 만든 뒤 내용을 바꿀 수 있는지를 뜻한다.

| 구분 | 예 | 변경할 때의 동작 |
| --- | --- | --- |
| Mutable | `list`, `dict`, `set`, `bytearray` | 같은 객체의 내용을 바꿀 수 있다. |
| Immutable | `int`, `float`, `complex`, `bool`, `str`, `tuple`, `frozenset`, `bytes` | 기존 객체의 값은 바뀌지 않는다. 다른 값을 쓰려면 이름을 다른 객체에 연결한다. |

사용자 정의 클래스의 인스턴스는 보통 가변 상태를 갖도록 만들지만, 설계에 따라 달라진다.


### 2.2 Immutable Types

불변 객체는 여러 곳에서 참조해도 그 값이 바뀌지 않는다. 값을 바꾸는 것처럼 보이는 연산도 기존 객체를 수정하는지, 다른 객체를 반환하는지 구분해서 보자.

`int`, `str` 등은 해시 가능하므로 딕셔너리 키로 쓸 수 있다. 다만 불변 컨테이너라고 항상 해시 가능한 것은 아니다. 예를 들어 List를 담은 Tuple은 해시할 수 없다.


````python
# Immutable 실험
print("=== Immutable Types ===")
print()

# Integer
x = 10
print(f"Integer:")
print(f"  x = 10,  id: {id(x)}")
x += 1
print(f"  x += 1,  id: {id(x)}  (새 객체!)")
print()

# String
s = "Hello"
print(f"String:")
print(f"  s = 'Hello',    id: {id(s)}")
s += " World"
print(f"  s += ' World',  id: {id(s)}  (새 객체!)")
print()

# Tuple
t = (1, 2, 3)
print(f"Tuple:")
print(f"  t = (1, 2, 3), id: {id(t)}")
# t[0] = 10  # 오류: TypeError!
print(f"  t[0] = 10은 불가능 (TypeError)")
t = (10, 2, 3)  # 새 튜플 생성
print(f"  t = (10, 2, 3), id: {id(t)}  (새 객체!)")
````

### 2.3 Mutable Types

`list`, `dict`, `set`은 객체를 새로 만들지 않고 내용을 수정할 수 있다. 같은 객체를 여러 이름이 가리키고 있다면 한쪽에서 바꾼 내용이 다른 쪽에서도 보인다.

이 세 타입은 해시할 수 없으므로 딕셔너리 키나 집합의 요소로 사용할 수 없다.


````python
# Mutable 실험
print("=== Mutable Types ===")
print()

# List
lst = [1, 2, 3]
print(f"List:")
print(f"  lst = [1, 2, 3],  id: {id(lst)}")
lst[0] = 10
print(f"  lst[0] = 10,      id: {id(lst)}  (같은 객체!)")
lst.append(4)
print(f"  lst.append(4),    id: {id(lst)}  (같은 객체!)")
print(f"  현재 lst: {lst}")
print()

# Dict
d = {'a': 1, 'b': 2}
print(f"Dict:")
print(f"  d = {{'a': 1, 'b': 2}},  id: {id(d)}")
d['c'] = 3
print(f"  d['c'] = 3,              id: {id(d)}  (같은 객체!)")
print(f"  현재 d: {d}")
print()

# Set
s = {1, 2, 3}
print(f"Set:")
print(f"  s = {{1, 2, 3}},  id: {id(s)}")
s.add(4)
print(f"  s.add(4),        id: {id(s)}  (같은 객체!)")
print(f"  현재 s: {s}")
````

### 2.4 Mutability의 영향

#### 1. Mutable Default Argument

기본 인자는 함수 정의 시 한 번 평가된다. 아래 함수는 호출할 때마다 같은 List를 사용한다.

```python
def func(lst=[]):
    lst.append(1)
    return lst
```

호출마다 새 List가 필요하다면 `None`을 기본값으로 두고 함수 안에서 만든다.

```python
def func(lst=None):
    if lst is None:
        lst = []
    lst.append(1)
    return lst
```


````python
# Default Argument 함정
print("=== Default Argument 함정 ===")
print()

# 호출 사이에 상태가 공유되는 코드


def add_item_bad(item, lst=[]):
  lst.append(item)
  return lst


print("위험한 함수:")
result1 = add_item_bad(1)
print(f"  add_item_bad(1): {result1}")
result2 = add_item_bad(2)
print(f"  add_item_bad(2): {result2}  ← 이전 값이 남아있음!")
print()

# 호출마다 새 객체를 만드는 코드


def add_item_good(item, lst=None):
  if lst is None:
    lst = []
    print(id(lst))
  lst.append(item)
  return lst


print("안전한 함수:")
result1 = add_item_good(1)
print(f"  add_item_good(1): {result1}")
result2 = add_item_good(2)
print(f"  add_item_good(2): {result2}  ← 올바름!")
````

#### 2. 얕은 복사 vs 깊은 복사

얕은 복사(Shallow Copy)는 새 컨테이너를 만들고 내부 요소의 참조를 복사한다. `list.copy()`, `dict.copy()`, `set.copy()`가 해당한다. 내부의 가변 객체는 원본과 공유한다.

`copy.deepcopy()`는 내부 객체도 재귀적으로 복사한다. 중첩 List 예제에서 어느 객체까지 원본과 공유하는지 `id()`로 확인해 보자.


````python
import sys

x = 100000

a = x
b = x

print(sys.getrefcount(x))

print(id(x), id(a), id(b))
````

````python
# 복사의 차이
import copy
print("=== 복사의 종류 ===")
print()

# 1. 참조 복사 (같은 객체)
original = [1, 2, 3]
ref_copy = original
ref_copy[0] = 10
print(f"참조 복사:")
print(f"  original: {original}")
print(f"  ref_copy: {ref_copy}")
print(f"  → 같은 객체를 가리킴!")
print()

# 2. 얕은 복사 (새 객체, 내부는 참조)
original = [[1, 2], [3, 4]]
shallow = original.copy()
shallow[0][0] = 10
print(f"얕은 복사:")
print(f"  original: {original}")
print(f"  shallow:  {shallow}")
print(f"  → 내부 리스트는 같은 객체!")
print()

# 3. 깊은 복사 (완전히 독립적)
original = [[1, 2], [3, 4]]
deep = copy.deepcopy(original)
deep[0][0] = 10
print(f"깊은 복사:")
print(f"  original: {original}")
print(f"  deep:     {deep}")
print(f"  → 완전히 독립적!")
````

#### 3. Tuple 안의 Mutable 객체

Tuple의 요소를 다른 객체로 바꿀 수는 없다. 다만 요소가 List 같은 가변 객체라면 그 객체의 내용은 수정할 수 있다.


````python
# Tuple 안의 Mutable
print("=== Tuple 안의 Mutable ===")
print()

# Tuple은 Immutable이지만...
t = ([1, 2], [3, 4])
print(f"t = ([1, 2], [3, 4])")
print(f"id(t): {id(t)}")
print()

# Tuple 자체는 변경 불가
# t[0] = [10, 20]  # 오류: TypeError!
print("t[0] = [10, 20]은 불가능")
print()

# 하지만 내부 리스트는 변경 가능!
t[0][0] = 10
print(f"t[0][0] = 10 실행")
print(f"t = {t}")
print(f"id(t): {id(t)}  (튜플 id는 동일)")
print()
print("→ Tuple은 불변이지만, 내부 Mutable 객체는 변경 가능!")
````

## 3. Type System

### 3.1 Dynamic Typing

Python에서는 변수에 타입을 선언하지 않고 객체를 연결한다. 같은 이름을 다른 타입의 객체에 다시 연결할 수 있으며, 사용할 수 없는 연산을 수행하면 실행 중에 오류가 발생한다.

```python
x = 42
x = "Hello"
```

Java의 `int x = 42;`처럼 변수 타입을 선언하는 방식과 비교해 보자.


````python
# Dynamic Typing
print("=== Dynamic Typing ===")
print()

# 같은 변수에 다른 타입
x = 42
print(f"x = 42,      type: {type(x)}")

x = "Hello"
print(f"x = 'Hello', type: {type(x)}")

x = [1, 2, 3]
print(f"x = [1,2,3], type: {type(x)}")
print()

# 함수도 동적 타입


def add(a, b):
  return a + b


print(f"add(10, 20) = {add(10, 20)}")
print(f"add('Hello', ' World') = {add('Hello', ' World')}")
print(f"add([1], [2, 3]) = {add([1], [2, 3])}")
````

### 3.2 Strong Typing

Python은 맞지 않는 타입 사이의 연산을 자동 변환으로 처리하기보다 오류로 알리는 경우가 많다. 예를 들어 `"10" + 5`는 `TypeError`가 발생한다. 숫자로 더하려면 `int("10") + 5`처럼 변환을 명시한다.

반면 JavaScript의 `"10" + 5`는 문자열 `"105"`가 된다. 같은 연산자라도 언어와 피연산자의 타입에 따라 동작이 다르다.


````python
# Strong Typing
print("=== Strong Typing ===")
print()

# 타입이 맞지 않으면 오류
try:
  result = "10" + 5
except TypeError as e:
  print(f"'10' + 5 → TypeError: {e}")
print()

# 명시적 변환 필요
result = int("10") + 5
print(f"int('10') + 5 = {result}  ✅")
print()

# Weak Typing 언어(JavaScript)와 비교
print("JavaScript (Weak Typing):")
print("  '10' + 5 = '105'  (문자열 결합)")
print("\nPython (Strong Typing):")
print("  '10' + 5 = TypeError  (명시적 변환 필요)")
````

### 3.3 Duck Typing

Duck Typing은 타입 이름보다 필요한 동작을 지원하는지에 따라 객체를 사용하는 방식이다. `obj.quack()`을 호출하는 코드는 `obj`의 구체적인 클래스 대신 `quack()`의 동작에 의존한다.

필요한 메서드가 없으면 실행 중에 오류가 발생한다. 메서드를 미리 확인하는 LBYL과 예상한 예외를 처리하는 EAFP를 비교해 보자.

```python
# 호출 전에 확인

if hasattr(obj, 'quack'):
    obj.quack()

# 호출 후 예상한 예외 처리

try:
    obj.quack()
except AttributeError:
    pass
```


````python
# Duck Typing
print("=== Duck Typing ===")
print()

# 다양한 "파일 같은" 객체


class FakeFile:
  def read(self):
    return "가짜 파일 내용"

  def write(self, data):
    print(f"가짜 파일에 쓰기: {data}")


def process_file(file_like):
  """파일처럼 동작하는 모든 객체 처리"""
  content = file_like.read()
  print(f"읽은 내용: {content}")
  file_like.write("새 데이터")


# 진짜 파일처럼 동작
fake = FakeFile()
process_file(fake)
print()

# 다른 예시: 리스트처럼 동작


class MyList:
  def __init__(self):
    self.items = []

  def append(self, item):
    self.items.append(item)

  def __len__(self):
    return len(self.items)


def add_items(list_like, items):
  """리스트처럼 동작하는 모든 객체에 아이템 추가"""
  for item in items:
    list_like.append(item)
  print(f"추가 후 길이: {len(list_like)}")


my_list = MyList()
add_items(my_list, [1, 2, 3])
````

### 3.4 Type Hints (Python 3.5+)

Type Hint는 인자와 반환값 등에 기대하는 타입을 적는 방법이다. IDE와 mypy 같은 정적 검사 도구가 활용하며, Python이 이 힌트에 맞춰 인자의 타입을 자동으로 검사하지는 않는다.

```python
def func(name: str, age: int) -> str:
    return f"{name} is {age} years old"

count: int = 0
names: list[str] = ["Alice", "Bob"]
```


````python
# Type Hints
print("=== Type Hints ===")
print()

# 함수에 타입 힌트


def add(a: int, b: int) -> int:
  return a + b


def greet(name: str) -> str:
  return f"Hello, {name}!"


# 타입 힌트는 강제가 아님!
result1 = add(10, 20)
result2 = add("Hello", " World")  # 오류 안 남! (힌트일 뿐)

print(f"add(10, 20) = {result1}")
print(f"add('Hello', ' World') = {result2}  (타입 힌트 무시)")
print()

# 변수에도 타입 힌트
age: int = 25
name: str = "Alice"
scores: list[int] = [90, 85, 95]

print(f"age: {age}, type: {type(age)}")
print(f"name: {name}, type: {type(name)}")
print(f"scores: {scores}, type: {type(scores)}")
````
