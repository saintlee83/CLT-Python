## 1. Class 기초 (50분)

### 1.1 Class란?

Class는 객체의 속성과 동작을 정의하고, Instance는 그 Class로 만든 객체다. 아래 `Person` 예제에서 Class 하나로 여러 Instance를 만들고 각자의 속성을 확인한다.

객체 지향 설계에서 자주 다루는 개념은 다음과 같다.

- 캡슐화: 데이터와 메서드를 묶고 사용할 인터페이스를 정한다.
- 상속: 기존 Class의 동작을 이어받아 확장한다.
- 다형성: 같은 인터페이스로 서로 다른 동작을 수행한다.
- 추상화: 사용에 필요한 기능을 드러내고 세부 구현을 감춘다.


````python
# 가장 기본적인 Class
print("=== 기본 Class 예시 ===")

class Dog:
    """강아지를 표현하는 클래스"""
    pass  # 아무 내용 없는 클래스

# Instance 생성
dog1 = Dog()
dog2 = Dog()

print(f"dog1: {dog1}")
print(f"dog2: {dog2}")
print(f"dog1과 dog2는 같은 객체? {dog1 is dog2}")
print(f"dog1의 타입: {type(dog1)}")
print(f"dog1은 Dog의 인스턴스? {isinstance(dog1, Dog)}")
````

### 1.2 `__init__`과 속성 초기화

`__init__`은 Instance를 초기화하는 메서드다. `self`는 해당 Instance를 가리키며, `obj.method()` 형태로 인스턴스 메서드를 호출할 때 자동으로 전달된다.

Instance Attribute는 인스턴스별로 저장하는 속성이고, Class Attribute는 Class에 저장해 여러 인스턴스에서 참조할 수 있는 속성이다. 아래 예제에서 각각 어디에 정의되는지 확인한다.


````python
# __init__을 사용한 Class
print("=== __init__ 생성자 ===")

class Dog:
    # Class Attribute (모든 인스턴스가 공유)
    species = "Canis familiaris"
    
    def __init__(self, name, age):
        # Instance Attribute (각 인스턴스마다 다름)
        self.name = name
        self.age = age
        print(f"{self.name} 생성됨!")

# Instance 생성
dog1 = Dog("바둑이", 3)
dog2 = Dog("코코", 5)

print(f"\ndog1: {dog1.name}, {dog1.age}살")
print(f"dog2: {dog2.name}, {dog2.age}살")
print(f"\nClass Attribute (공유): {dog1.species} = {dog2.species}")
print(f"Instance Attribute (개별): {dog1.name} ≠ {dog2.name}")
````

````python
# Class Attribute와 Instance Attribute의 차이
print("\n=== Attribute 차이 ===")

class Counter:
    count = 0  # Class Attribute
    
    def __init__(self):
        Counter.count += 1  # Class Attribute 수정
        self.id = Counter.count  # Instance Attribute

c1 = Counter()
c2 = Counter()
c3 = Counter()

print(f"총 생성된 Counter: {Counter.count}")
print(f"c1.id: {c1.id}")
print(f"c2.id: {c2.id}")
print(f"c3.id: {c3.id}")
````

### 1.3 Method의 종류

#### 1. Instance Method (인스턴스 메서드)

- 정의: `self`를 첫 번째 매개변수로 받는 메서드
- 용도: Instance의 속성에 접근하거나 수정
- **호출**: `instance.method()`

#### 2. Class Method (클래스 메서드)

- 정의: `@classmethod` 데코레이터 사용, `cls`를 첫 번째 매개변수로 받음
- 용도: Class Attribute에 접근, Alternative Constructor (대체 생성자)
- **호출**: `Class.method()` 또는 `instance.method()`

#### 3. Static Method (정적 메서드)

- 정의: `@staticmethod` 데코레이터 사용, `self`나 `cls` 없음
- 용도: Class와 관련은 있지만 Instance나 Class Attribute에 접근 불필요한 경우
- **호출**: `Class.method()` 또는 `instance.method()`


````python
# Method의 종류
print("=== Method 종류 ===")

class Person:
    population = 0  # Class Attribute
    
    def __init__(self, name, age):
        self.name = name
        self.age = age
        Person.population += 1
    
    # 1. Instance Method
    def introduce(self):
        return f"안녕하세요, 저는 {self.name}이고 {self.age}살입니다."
    
    def birthday(self):
        self.age += 1
        print(f"{self.name}의 생일! 이제 {self.age}살")
    
    # 2. Class Method (대체 생성자)
    @classmethod
    def from_birth_year(cls, name, birth_year):
        age = 2025 - birth_year
        return cls(name, age)
    
    @classmethod
    def get_population(cls):
        return f"현재 인구: {cls.population}명"
    
    # 3. Static Method
    @staticmethod
    def is_adult(age):
        return age >= 18
    
    @staticmethod
    def calculate_age(birth_year):
        return 2025 - birth_year

# Instance Method 사용
p1 = Person("철수", 25)
print(p1.introduce())
p1.birthday()

# Class Method 사용 (대체 생성자)
p2 = Person.from_birth_year("영희", 2000)
print(p2.introduce())

# Class Method 사용 (Class Attribute 접근)
print(Person.get_population())

# Static Method 사용
print(f"\n20살은 성인? {Person.is_adult(20)}")
print(f"2000년생의 나이: {Person.calculate_age(2000)}살")
````

````python
# Method 종류 비교 정리
print("\n=== Method 비교 ===")

class Example:
    class_var = "클래스 변수"
    
    def __init__(self, value):
        self.instance_var = value
    
    def instance_method(self):
        # self를 통해 Instance와 Class에 모두 접근 가능
        return f"Instance: {self.instance_var}, Class: {self.class_var}"
    
    @classmethod
    def class_method(cls):
        # cls를 통해 Class에만 접근 가능
        return f"Class: {cls.class_var}"
        # return f"Instance: {cls.instance_var}"  # 에러!
    
    @staticmethod
    def static_method(param):
        # self나 cls 없이 독립적으로 동작
        return f"Static: {param}"
        # return f"Instance: {self.instance_var}"  # 에러!

ex = Example("인스턴스 변수")

print("Instance Method:", ex.instance_method())
print("Class Method:", Example.class_method())
print("Static Method:", Example.static_method("파라미터"))
````

### 언제 어떤 Method를 사용할까?

| Method 종류 | 언제 사용? | 예시 |
|------------|----------|------|
| **Instance Method** | Instance의 상태(속성)를 읽거나 수정할 때 | `obj.get_name()`, `obj.set_age(25)` |
| **Class Method** | 1. 대체 생성자가 필요할 때<br>2. Class Attribute를 다룰 때 | `Person.from_birth_year(2000)`<br>`Person.get_count()` |
| **Static Method** | Class와 관련은 있지만<br>Instance/Class 속성이 불필요할 때 | `Math.is_even(4)`<br>`Validator.check_email()` |


## 2. Special Methods (Dunder Methods) (50분)

### 2.1 Special Methods란?

`__len__`, `__add__`처럼 이름 앞뒤에 밑줄 두 개가 붙는 메서드다. 이 메서드를 구현하면 직접 만든 객체도 `len()`, `+`, `in`, `with` 같은 Python의 연산과 문법에 참여할 수 있다.

이번 절에서는 문자열 표현, 비교·산술 연산, 컨테이너 접근, 함수 호출, Context Manager에 해당하는 메서드를 작성한다.


### 2.2 기본 Special Methods

`__str__`은 `str(obj)`나 `print(obj)`에서 사용할 문자열을, `__repr__`은 `repr(obj)`나 대화형 환경에서 객체를 확인할 때 사용할 표현을 만든다. `__str__`이 없으면 `__repr__`이 대신 쓰인다.

`__repr__`에는 객체를 구별하고 상태를 확인하는 데 필요한 정보를 담는다. 가능하다면 객체를 다시 만들 수 있는 표현을 사용하되, 표현만으로 재생성하기 어려운 객체도 있다.


````python
# __str__과 __repr__ 비교
print("=== __str__ vs __repr__ ===")

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def __str__(self):
        # 사용자 친화적 표현
        return f"{self.name} ({self.age}살)"
    
    def __repr__(self):
        # 개발자용, 객체 재생성 가능한 표현
        return f"Person('{self.name}', {self.age})"

p = Person("김철수", 25)

print("str(p):", str(p))      # __str__ 호출
print("repr(p):", repr(p))    # __repr__ 호출
print("print(p):", end=" ")  
print(p)                       # __str__ 호출

# 콘솔에서 변수만 입력하면 __repr__ 호출
print("\n콘솔에서 p 입력 시:")
print(repr(p))
````

````python
# __str__만 있을 때 vs 둘 다 있을 때
print("\n=== __str__과 __repr__ 유무 비교 ===")

class OnlyStr:
    def __init__(self, value):
        self.value = value
    
    def __str__(self):
        return f"Value: {self.value}"

class OnlyRepr:
    def __init__(self, value):
        self.value = value
    
    def __repr__(self):
        return f"OnlyRepr({self.value})"

obj1 = OnlyStr(10)
obj2 = OnlyRepr(20)

print("OnlyStr:")
print(f"  print(): {obj1}")
print(f"  repr(): {repr(obj1)}")

print("\nOnlyRepr:")
print(f"  print(): {obj2}")
print(f"  repr(): {repr(obj2)}")
````

### 2.3 비교 연산자 Special Methods

#### Comparison Operators

| 연산자 | Special Method | 설명 |
|--------|---------------|------|
| `==` | `__eq__(self, other)` | Equal |
| `!=` | `__ne__(self, other)` | Not equal |
| `<` | `__lt__(self, other)` | Less than |
| `>` | `__gt__(self, other)` | Greater than |
| `<=` | `__le__(self, other)` | Less than or equal |
| `>=` | `__ge__(self, other)` | Greater than or equal |

**Tip**: `__eq__`와 `__lt__`만 구현하면 나머지는 자동으로 처리 가능 (`functools.total_ordering` 데코레이터 사용)


````python
# 비교 연산자 구현
print("=== 비교 연산자 Special Methods ===")

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def __eq__(self, other):
        """== 연산자"""
        if not isinstance(other, Point):
            return False
        return self.x == other.x and self.y == other.y
    
    def __lt__(self, other):
        """< 연산자 (원점으로부터의 거리 비교)"""
        return (self.x**2 + self.y**2) < (other.x**2 + other.y**2)
    
    def __gt__(self, other):
        """> 연산자"""
        return (self.x**2 + self.y**2) > (other.x**2 + other.y**2)
    
    def __repr__(self):
        return f"Point({self.x}, {self.y})"

p1 = Point(3, 4)  # 거리: 5
p2 = Point(3, 4)  # 거리: 5
p3 = Point(1, 2)  # 거리: 2.24

print(f"p1 == p2: {p1 == p2}")
print(f"p1 == p3: {p1 == p3}")
print(f"p1 > p3: {p1 > p3}")
print(f"p3 < p1: {p3 < p1}")
````

````python
# functools.total_ordering을 사용한 자동 비교
from functools import total_ordering

print("\n=== total_ordering 데코레이터 ===")

@total_ordering
class Student:
    def __init__(self, name, grade):
        self.name = name
        self.grade = grade
    
    def __eq__(self, other):
        return self.grade == other.grade
    
    def __lt__(self, other):
        return self.grade < other.grade
    
    def __repr__(self):
        return f"Student('{self.name}', {self.grade})"

s1 = Student("철수", 85)
s2 = Student("영희", 90)
s3 = Student("민수", 85)

print(f"s1 == s3: {s1 == s3}")
print(f"s1 < s2: {s1 < s2}")
print(f"s2 > s1: {s2 > s1}")
print(f"s1 <= s3: {s1 <= s3}")  # __eq__와 __lt__로 자동 생성!
print(f"s2 >= s1: {s2 >= s1}")  # 자동 생성!
````

### 2.4 산술 연산자 Special Methods

#### Arithmetic Operators

| 연산자 | Special Method | 설명 |
|--------|---------------|------|
| `+` | `__add__(self, other)` | 덧셈 |
| `-` | `__sub__(self, other)` | 뺄셈 |
| `*` | `__mul__(self, other)` | 곱셈 |
| `/` | `__truediv__(self, other)` | 나눗셈 |
| `//` | `__floordiv__(self, other)` | 몫 |
| `%` | `__mod__(self, other)` | 나머지 |
| `**` | `__pow__(self, other)` | 거듭제곱 |


````python
# 산술 연산자 구현 (Vector 예제)
print("=== 산술 연산자 Special Methods ===")

class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def __repr__(self):
        return f"Vector({self.x}, {self.y})"
    
    def __add__(self, other):
        """벡터 덧셈"""
        return Vector(self.x + other.x, self.y + other.y)
    
    def __sub__(self, other):
        """벡터 뺄셈"""
        return Vector(self.x - other.x, self.y - other.y)
    
    def __mul__(self, scalar):
        """스칼라 곱셈"""
        return Vector(self.x * scalar, self.y * scalar)
    
    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

v1 = Vector(1, 2)
v2 = Vector(3, 4)

print(f"v1 + v2 = {v1 + v2}")
print(f"v2 - v1 = {v2 - v1}")
print(f"v1 * 3 = {v1 * 3}")
print(f"v1 == v2: {v1 == v2}")
````

### 2.5 Container Special Methods

#### Container Protocol

| Method | 용도 | 호출 방법 |
|--------|------|----------|
| `__len__(self)` | 길이 반환 | `len(obj)` |
| `__getitem__(self, key)` | 항목 가져오기 | `obj[key]` |
| `__setitem__(self, key, value)` | 항목 설정 | `obj[key] = value` |
| `__delitem__(self, key)` | 항목 삭제 | `del obj[key]` |
| `__contains__(self, item)` | 포함 여부 | `item in obj` |
| `__iter__(self)` | 반복자 반환 | `for x in obj` |


````python
# Container Special Methods
print("=== Container Special Methods ===")

class MyList:
    def __init__(self):
        self.items = []
    
    def __len__(self):
        """len() 함수 지원"""
        return len(self.items)
    
    def __getitem__(self, index):
        """인덱싱 지원"""
        return self.items[index]
    
    def __setitem__(self, index, value):
        """항목 할당 지원"""
        self.items[index] = value
    
    def __contains__(self, item):
        """in 연산자 지원"""
        return item in self.items
    
    def append(self, item):
        self.items.append(item)
    
    def __repr__(self):
        return f"MyList({self.items})"

my_list = MyList()
my_list.append(1)
my_list.append(2)
my_list.append(3)

print(f"my_list: {my_list}")
print(f"len(my_list): {len(my_list)}")
print(f"my_list[0]: {my_list[0]}")
print(f"2 in my_list: {2 in my_list}")
print(f"5 in my_list: {5 in my_list}")

my_list[0] = 10
print(f"\nAfter my_list[0] = 10: {my_list}")
````

### 2.6 기타 유용한 Special Methods

#### `__call__`: 객체를 함수처럼 호출

- 객체를 함수처럼 사용할 수 있게 함
- `obj(args)`처럼 호출 가능

#### `__enter__`, `__exit__`: Context Manager

- `with` 문에서 사용
- 리소스 관리 (파일, 네트워크 연결 등)


````python
# __call__: 객체를 함수처럼 호출
print("=== __call__ 메서드 ===")

class Multiplier:
    def __init__(self, factor):
        self.factor = factor
    
    def __call__(self, x):
        """객체를 함수처럼 호출 가능"""
        return x * self.factor

# 2배 하는 객체
double = Multiplier(2)
# 3배 하는 객체
triple = Multiplier(3)

print(f"double(5) = {double(5)}")
print(f"triple(5) = {triple(5)}")
print(f"double(10) = {double(10)}")
````

````python
# Context Manager (__enter__, __exit__)
print("\n=== Context Manager ===")

class FileManager:
    def __init__(self, filename, mode):
        self.filename = filename
        self.mode = mode
        self.file = None
    
    def __enter__(self):
        """with 블록 진입 시 호출"""
        print(f"파일 열기: {self.filename}")
        self.file = open(self.filename, self.mode)
        return self.file
    
    def __exit__(self, exc_type, exc_val, exc_tb):
        """with 블록 종료 시 호출 (에러 발생 시에도 호출됨)"""
        print(f"파일 닫기: {self.filename}")
        if self.file:
            self.file.close()
        return False  # 예외를 re-raise

# with 문 사용
with FileManager('/tmp/test.txt', 'w') as f:
    f.write("Hello, World!")
    print("파일 작성 중...")

print("\nwith 블록 종료 (자동으로 파일이 닫힘)")
````

## 3. 상속과 다형성 (50분)

### 3.1 상속(Inheritance)

자식 Class는 부모 Class의 속성과 메서드를 이어받고, 필요한 부분을 추가하거나 재정의할 수 있다. 부모는 Base Class 또는 Superclass, 자식은 Derived Class 또는 Subclass라고도 한다.

```python
class ChildClass(ParentClass):
    pass
```


````python
# 기본 상속
print("=== 기본 상속 ===")

class Animal:
    """부모 클래스"""
    def __init__(self, name):
        self.name = name
    
    def speak(self):
        return f"{self.name}이(가) 소리를 냅니다."
    
    def move(self):
        return f"{self.name}이(가) 움직입니다."

class Dog(Animal):
    """자식 클래스"""
    def speak(self):  # 메서드 오버라이딩
        return f"{self.name}: 멍멍!"

class Cat(Animal):
    """자식 클래스"""
    def speak(self):  # 메서드 오버라이딩
        return f"{self.name}: 야옹~"

# 부모 클래스 사용
animal = Animal("동물")
print(animal.speak())
print(animal.move())

# 자식 클래스 사용
dog = Dog("바둑이")
print(dog.speak())      # 오버라이딩된 메서드
print(dog.move())       # 상속받은 메서드

cat = Cat("나비")
print(cat.speak())      # 오버라이딩된 메서드
print(cat.move())       # 상속받은 메서드
````

### 3.2 `super()` 함수

`super()`는 MRO에서 현재 Class 다음에 있는 구현을 찾을 때 사용한다. 단일 상속에서는 보통 부모의 초기화나 메서드를 이어서 호출한다. 다중 상속에서는 실제 호출 순서를 MRO로 확인해야 한다.

```python
super().__init__(args)
super().method_name(args)
```


````python
# super() 사용
print("\n=== super() 함수 ===")

class Animal:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        print(f"Animal 생성: {name}")
    
    def info(self):
        return f"{self.name} ({self.age}살)"

class Dog(Animal):
    def __init__(self, name, age, breed):
        # 부모 클래스의 __init__ 호출
        super().__init__(name, age)
        self.breed = breed
        print(f"Dog 생성: {breed}")
    
    def info(self):
        # 부모 클래스의 메서드를 확장
        base_info = super().info()
        return f"{base_info}, 품종: {self.breed}"

dog = Dog("바둑이", 3, "진돗개")
print(f"\n정보: {dog.info()}")
````

### 3.3 다중 상속(Multiple Inheritance)

Python Class는 여러 부모 Class를 상속할 수 있다. 같은 이름의 메서드가 있으면 MRO(Method Resolution Order)에 따라 찾는다.

`ClassName.__mro__` 또는 `ClassName.mro()`로 순서를 확인한다. Python은 C3 Linearization을 사용해 자식이 부모보다 앞서고, 선언한 부모 순서를 유지하는 일관된 순서를 만든다.


````python
# 다중 상속
print("=== 다중 상속 ===")

class Flyable:
    def fly(self):
        return "날 수 있어요!"

class Swimmable:
    def swim(self):
        return "헤엄칠 수 있어요!"

class Duck(Flyable, Swimmable):
    def __init__(self, name):
        self.name = name
    
    def speak(self):
        return f"{self.name}: 꽥꽥!"

duck = Duck("도널드")
print(duck.speak())
print(duck.fly())    # Flyable에서 상속
print(duck.swim())   # Swimmable에서 상속
````

````python
# MRO (Method Resolution Order)
print("\n=== MRO 예제 ===")

class A:
    def method(self):
        print("A의 method")

class B(A):
    def method(self):
        print("B의 method")
        super().method()

class C(A):
    def method(self):
        print("C의 method")
        super().method()

class D(B, C):
    def method(self):
        print("D의 method")
        super().method()

# MRO 확인
print("D의 MRO:")
for cls in D.__mro__:
    print(f"  - {cls.__name__}")

print("\nmethod 호출 순서:")
d = D()
d.method()
# 출력: D → B → C → A
````

### 3.4 다형성(Polymorphism)

다형성은 같은 인터페이스로 서로 다른 동작을 수행하는 것이다. 자식 Class가 메서드를 재정의하는 Method Overriding과, 필요한 동작을 제공하는 객체를 받는 Duck Typing으로 살펴볼 수 있다.

아래 예제에서 호출 코드는 그대로 두고 전달하는 객체만 바꿔보자.


````python
# 다형성 예제
print("=== 다형성 (Polymorphism) ===")

class Animal:
    def speak(self):
        raise NotImplementedError("서브클래스에서 구현해야 함")

class Dog(Animal):
    def speak(self):
        return "멍멍!"

class Cat(Animal):
    def speak(self):
        return "야옹~"

class Cow(Animal):
    def speak(self):
        return "음메~"

# 다형성: 같은 메서드(speak)를 호출하지만 다른 동작
def animal_sound(animal):
    """Animal 타입이면 무엇이든 받을 수 있음"""
    print(animal.speak())

animals = [Dog(), Cat(), Cow()]

for animal in animals:
    animal_sound(animal)
````

````python
# Duck Typing 예제
print("\n=== Duck Typing ===")

class Duck:
    def speak(self):
        return "꽥꽥!"

class Person:
    def speak(self):
        return "안녕하세요!"

class Robot:
    def speak(self):
        return "삐빅! 삐빅!"

# Duck Typing: speak 메서드만 있으면 모두 사용 가능
# Animal을 상속받지 않아도 됨!
def make_it_speak(thing):
    """speak 메서드가 있기만 하면 작동"""
    print(thing.speak())

things = [Duck(), Person(), Robot()]

for thing in things:
    make_it_speak(thing)

print("\n→ 상속 관계가 없어도 같은 인터페이스면 사용 가능!")
````

### 3.5 Abstract Base Class (추상 기본 클래스)

`abc` 모듈의 `ABC`와 `@abstractmethod`로 자식 Class가 제공해야 할 메서드를 명시한다. 추상 메서드가 남아 있는 Class는 인스턴스를 만들 수 없다. 아래 예제에서 메서드 구현 전후를 비교해 보자.


````python
# Abstract Base Class
from abc import ABC, abstractmethod

print("=== Abstract Base Class ===")

class Shape(ABC):
    """추상 기본 클래스"""
    
    @abstractmethod
    def area(self):
        """넓이를 계산하는 추상 메서드"""
        pass
    
    @abstractmethod
    def perimeter(self):
        """둘레를 계산하는 추상 메서드"""
        pass

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height
    
    def area(self):
        return self.width * self.height
    
    def perimeter(self):
        return 2 * (self.width + self.height)

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        return 3.14159 * self.radius ** 2
    
    def perimeter(self):
        return 2 * 3.14159 * self.radius

# Shape는 직접 인스턴스화 불가
# shape = Shape()  # TypeError 발생!

# 자식 클래스는 인스턴스화 가능
rect = Rectangle(5, 3)
circle = Circle(4)

print(f"직사각형 넓이: {rect.area()}")
print(f"직사각형 둘레: {rect.perimeter()}")
print(f"원 넓이: {circle.area():.2f}")
print(f"원 둘레: {circle.perimeter():.2f}")
````

## 4. Encapsulation (캡슐화)

### 4.1 캡슐화와 이름 관례

캡슐화는 데이터와 동작을 묶고, 외부에서 사용할 인터페이스를 정하는 일이다. Python에서는 이름으로 내부 구현임을 알리는 관례를 사용한다.

- `name`: 외부에서 사용하는 공개 이름
- `_name`: 내부 구현임을 나타내는 관례
- `__name`: Name Mangling으로 클래스별 이름 충돌을 줄이는 이름

`_name`은 접근을 강제로 막지 않으며, `__name`도 다른 언어의 접근 제한자와 같은 보안 장치는 아니다.


````python
# 접근 제어자
print("=== 접근 제어자 ===")

class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner           # public
        self._balance = balance      # protected (관례)
        self.__pin = "1234"          # private (Name Mangling)
    
    def deposit(self, amount):
        """입금"""
        if amount > 0:
            self._balance += amount
            print(f"{amount}원 입금 완료")
    
    def withdraw(self, amount, pin):
        """출금"""
        if pin != self.__pin:
            print("PIN이 올바르지 않습니다")
            return
        
        if amount > self._balance:
            print("잔액이 부족합니다")
            return
        
        self._balance -= amount
        print(f"{amount}원 출금 완료")
    
    def get_balance(self):
        """잔액 조회"""
        return self._balance

account = BankAccount("김철수", 10000)

# public 접근 - 가능
print(f"예금주: {account.owner}")

# protected 접근 - 가능하지만 권장하지 않음
print(f"잔액 (직접 접근): {account._balance}")

# private 접근 - Name Mangling으로 인해 불가
# print(account.__pin)  # AttributeError

# 메서드를 통한 접근 - 권장
print(f"잔액 (메서드): {account.get_balance()}원")
account.deposit(5000)
account.withdraw(3000, "1234")
````

````python
# Name Mangling 확인
print("\n=== Name Mangling ===")

class Example:
    def __init__(self):
        self.__private = "Private 변수"

ex = Example()

# 직접 접근 불가
try:
    print(ex.__private)
except AttributeError as e:
    print(f"에러: {e}")

# Name Mangling된 이름으로는 접근 가능 (권장하지 않음!)
print(f"\nName Mangling된 접근: {ex._Example__private}")
print("\n→ Python에는 진정한 private이 없음!")
````

### 4.2 Property (프로퍼티)

#### Property란?

- 정의: 메서드를 속성처럼 사용할 수 있게 하는 기능
- 장점:
  1. Getter/Setter를 간결하게 구현
  2. 값 검증 가능
  3. 계산된 속성 제공
  4. API 호환성 유지

**사용 방법**:
```python
@property
def attribute(self):      # Getter
    return self._attribute

@attribute.setter
def attribute(self, value):  # Setter
    self._attribute = value

@attribute.deleter
def attribute(self):      # Deleter
    del self._attribute
```


````python
# Property 기본 사용
print("=== Property 기본 ===")

class Temperature:
    def __init__(self, celsius):
        self._celsius = celsius
    
    @property
    def celsius(self):
        """섭씨 온도 Getter"""
        print("celsius getter 호출")
        return self._celsius
    
    @celsius.setter
    def celsius(self, value):
        """섭씨 온도 Setter (값 검증 포함)"""
        print(f"celsius setter 호출: {value}")
        if value < -273.15:
            raise ValueError("절대 영도보다 낮을 수 없습니다")
        self._celsius = value
    
    @property
    def fahrenheit(self):
        """화씨 온도 (계산된 속성)"""
        return self._celsius * 9/5 + 32
    
    @fahrenheit.setter
    def fahrenheit(self, value):
        """화씨 온도 설정"""
        self._celsius = (value - 32) * 5/9

# Property를 속성처럼 사용
temp = Temperature(25)
print(f"현재 온도: {temp.celsius}°C")  # Getter 호출
print(f"화씨 온도: {temp.fahrenheit:.1f}°F")

# Setter 사용
temp.celsius = 30  # Setter 호출
print(f"변경된 온도: {temp.celsius}°C")

# 값 검증
try:
    temp.celsius = -300  # 절대 영도보다 낮음
except ValueError as e:
    print(f"\n에러: {e}")
````

````python
# Property의 실용적인 예제
print("\n=== Property 실용 예제 ===")

class Person:
    def __init__(self, first_name, last_name, age):
        self.first_name = first_name
        self.last_name = last_name
        self._age = age
    
    @property
    def full_name(self):
        """계산된 속성: 전체 이름"""
        return f"{self.first_name} {self.last_name}"
    
    @full_name.setter
    def full_name(self, name):
        """전체 이름 설정 (자동으로 분리)"""
        parts = name.split()
        if len(parts) != 2:
            raise ValueError("이름은 '성 이름' 형식이어야 합니다")
        self.first_name, self.last_name = parts
    
    @property
    def age(self):
        """나이 Getter"""
        return self._age
    
    @age.setter
    def age(self, value):
        """나이 Setter (0-150 검증)"""
        if not 0 <= value <= 150:
            raise ValueError("나이는 0-150 사이여야 합니다")
        self._age = value
    
    @property
    def is_adult(self):
        """계산된 속성: 성인 여부"""
        return self._age >= 18

person = Person("김", "철수", 25)
print(f"이름: {person.full_name}")
print(f"나이: {person.age}")
print(f"성인? {person.is_adult}")

# Setter 사용
person.full_name = "이 영희"
print(f"\n변경된 이름: {person.full_name}")
print(f"성: {person.first_name}, 이름: {person.last_name}")

person.age = 15
print(f"\n변경된 나이: {person.age}")
print(f"성인? {person.is_adult}")
````

### 4.3 Property vs Getter/Setter

#### Getter/Setter 방식 (Java 스타일)

```python
class OldStyle:
    def get_value(self):
        return self._value

    def set_value(self, value):
        self._value = value

# 사용

obj.set_value(10)
print(obj.get_value())
```

#### Property 방식 (Pythonic)

```python
class NewStyle:
    @property
    def value(self):
        return self._value

    @value.setter
    def value(self, value):
        self._value = value

# 사용

obj.value = 10
print(obj.value)
```

**Property의 장점**:
1. 더 간결하고 직관적
2. 속성처럼 사용 가능
3. 기존 코드 호환성 유지 (공개 속성 → Property 전환 시)


````python
# Getter/Setter vs Property 비교
print("=== Getter/Setter vs Property ===")

# 1. Getter/Setter 방식
class OldStyleCircle:
    def __init__(self, radius):
        self._radius = radius
    
    def get_radius(self):
        return self._radius
    
    def set_radius(self, radius):
        if radius < 0:
            raise ValueError("반지름은 0 이상이어야 합니다")
        self._radius = radius
    
    def get_area(self):
        return 3.14159 * self._radius ** 2

# 2. Property 방식
class NewStyleCircle:
    def __init__(self, radius):
        self._radius = radius
    
    @property
    def radius(self):
        return self._radius
    
    @radius.setter
    def radius(self, radius):
        if radius < 0:
            raise ValueError("반지름은 0 이상이어야 합니다")
        self._radius = radius
    
    @property
    def area(self):
        return 3.14159 * self._radius ** 2

print("Old Style (Getter/Setter):")
c1 = OldStyleCircle(5)
print(f"  반지름: {c1.get_radius()}")
print(f"  넓이: {c1.get_area():.2f}")
c1.set_radius(10)
print(f"  변경 후 반지름: {c1.get_radius()}")

print("\nNew Style (Property):")
c2 = NewStyleCircle(5)
print(f"  반지름: {c2.radius}")
print(f"  넓이: {c2.area:.2f}")
c2.radius = 10
print(f"  변경 후 반지름: {c2.radius}")

print("\n→ Property가 더 간결하고 Pythonic!")
````

## 종합 예제

### 은행 계좌 시스템 (클래스, 상속, 프로퍼티 활용)


````python
# 종합 예제: 은행 계좌 시스템
from abc import ABC, abstractmethod
from datetime import datetime

print("=== 은행 계좌 시스템 (종합 예제) ===")

class Account(ABC):
    """계좌 추상 기본 클래스"""
    
    account_count = 0  # Class Attribute
    
    def __init__(self, owner, balance=0):
        self.owner = owner
        self._balance = balance
        self.__pin = "1234"  # Private
        Account.account_count += 1
        self.account_number = f"ACC{Account.account_count:04d}"
        self.transactions = []
    
    # Property
    @property
    def balance(self):
        return self._balance
    
    # Abstract Method
    @abstractmethod
    def calculate_interest(self):
        pass
    
    # Instance Method
    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("입금액은 0보다 커야 합니다")
        self._balance += amount
        self._add_transaction("입금", amount)
        return self._balance
    
    def withdraw(self, amount, pin):
        if pin != self.__pin:
            raise ValueError("PIN이 올바르지 않습니다")
        if amount > self._balance:
            raise ValueError("잔액이 부족합니다")
        self._balance -= amount
        self._add_transaction("출금", -amount)
        return self._balance
    
    def _add_transaction(self, type_, amount):
        """Protected Method: 거래 내역 추가"""
        self.transactions.append({
            'type': type_,
            'amount': amount,
            'balance': self._balance,
            'time': datetime.now().strftime('%Y-%m-%d %H:%M:%S')
        })
    
    # Class Method
    @classmethod
    def get_account_count(cls):
        return cls.account_count
    
    # Special Methods
    def __str__(self):
        return f"{self.owner}의 {self.account_number} (잔액: {self._balance:,}원)"
    
    def __repr__(self):
        return f"{self.__class__.__name__}('{self.owner}', {self._balance})"
    
    def __eq__(self, other):
        return self.account_number == other.account_number
    
    def __lt__(self, other):
        return self._balance < other._balance

class SavingsAccount(Account):
    """예금 계좌"""
    
    def __init__(self, owner, balance=0, interest_rate=0.02):
        super().__init__(owner, balance)
        self.interest_rate = interest_rate
    
    def calculate_interest(self):
        """이자 계산 및 적용"""
        interest = self._balance * self.interest_rate
        self._balance += interest
        self._add_transaction("이자", interest)
        return interest

class CheckingAccount(Account):
    """입출금 계좌 (마이너스 통장)"""
    
    def __init__(self, owner, balance=0, overdraft_limit=1000000):
        super().__init__(owner, balance)
        self.overdraft_limit = overdraft_limit
    
    def calculate_interest(self):
        """마이너스 잔액에 대한 이자 부과"""
        if self._balance < 0:
            interest = abs(self._balance) * 0.05  # 5% 이자 부과
            self._balance -= interest
            self._add_transaction("이자 부과", -interest)
            return -interest
        return 0
    
    def withdraw(self, amount, pin):
        """오버라이딩: 마이너스 한도 확인"""
        if amount > self._balance + self.overdraft_limit:
            raise ValueError("마이너스 한도를 초과했습니다")
        return super().withdraw(amount, pin)

# 사용 예시
print("\n1. 계좌 생성")
savings = SavingsAccount("김철수", 1000000, 0.03)
checking = CheckingAccount("이영희", 500000, 2000000)

print(f"예금 계좌: {savings}")
print(f"입출금 계좌: {checking}")
print(f"총 계좌 수: {Account.get_account_count()}")

print("\n2. 입출금")
savings.deposit(500000)
print(f"입금 후: {savings}")

checking.withdraw(700000, "1234")
print(f"출금 후: {checking}")

print("\n3. 이자 계산")
interest = savings.calculate_interest()
print(f"예금 이자: {interest:,.0f}원")
print(f"이자 후: {savings}")

print("\n4. 계좌 비교")
print(f"savings > checking: {savings > checking}")

print("\n5. 거래 내역 (최근 3건)")
for trans in savings.transactions[-3:]:
    print(f"  {trans['time']}: {trans['type']} {trans['amount']:,.0f}원 (잔액: {trans['balance']:,.0f}원)")
````
